# Atlanta business snapshot

These charts use Atlanta metro data already in this repository: BLS QCEW fourth-quarter data for 2021–2025, aggregated to the project's fixed 29-county metro definition. This is metro-level context, not City of Atlanta-only data.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# Works when the notebook is opened from this Kelly folder.
data_path = Path('../Artem/data/nine_metro_models/metro_year_qcew_2021_2025.csv')
data = pd.read_csv(data_path)
atlanta = data.loc[data['metro'].eq('Atlanta')].sort_values('year').copy()
atlanta[['year', 'employment_q4_december', 'establishments_q4', 'avg_weekly_wage']]
```

## Private employment
December employment count by year.

```python
fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(atlanta['year'], atlanta['employment_q4_december'], marker='o', linewidth=3, color='#2F80ED')
ax.set_title('Atlanta Metro Private Employment, 2021–2025', loc='left', weight='bold')
ax.set_xlabel('Year'); ax.set_ylabel('Private jobs')
ax.yaxis.set_major_formatter(lambda x, pos: f'{x/1_000_000:.2f}M')
ax.grid(axis='y', alpha=.25); ax.spines[['top', 'right']].set_visible(False)
for x, y in zip(atlanta['year'], atlanta['employment_q4_december']):
    ax.annotate(f'{y/1_000_000:.3f}M', (x, y), xytext=(0, 9), textcoords='offset points', ha='center', fontsize=9)
plt.tight_layout(); plt.show()
print(f"Change, 2021–2025: {atlanta['employment_q4_december'].iloc[-1] - atlanta['employment_q4_december'].iloc[0]:,.0f} jobs ({(atlanta['employment_q4_december'].iloc[-1] / atlanta['employment_q4_december'].iloc[0] - 1) * 100:.1f}%).")
```

## Private establishments
Fourth-quarter establishment count by year.

```python
fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.bar(atlanta['year'].astype(str), atlanta['establishments_q4'], color='#27AE60', width=.62)
ax.set_title('Atlanta Metro Private Establishments, 2021–2025', loc='left', weight='bold')
ax.set_xlabel('Year'); ax.set_ylabel('Establishments')
ax.yaxis.set_major_formatter(lambda x, pos: f'{x/1000:.0f}k')
ax.grid(axis='y', alpha=.25); ax.set_axisbelow(True); ax.spines[['top', 'right']].set_visible(False)
ax.bar_label(bars, labels=[f'{v:,.0f}' for v in atlanta['establishments_q4']], padding=4, fontsize=9)
plt.tight_layout(); plt.show()
print(f"Change, 2021–2025: {atlanta['establishments_q4'].iloc[-1] - atlanta['establishments_q4'].iloc[0]:,.0f} establishments ({(atlanta['establishments_q4'].iloc[-1] / atlanta['establishments_q4'].iloc[0] - 1) * 100:.1f}%).")
```

## Average weekly wages
Nominal Q4 average weekly wage; the values are not adjusted for inflation.

```python
fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(atlanta['year'], atlanta['avg_weekly_wage'], marker='o', linewidth=3, color='#F2994A')
ax.set_title('Atlanta Metro Average Weekly Wage, 2021–2025', loc='left', weight='bold')
ax.set_xlabel('Year'); ax.set_ylabel('Dollars per week')
ax.yaxis.set_major_formatter(lambda x, pos: f'${x:,.0f}')
ax.grid(axis='y', alpha=.25); ax.spines[['top', 'right']].set_visible(False)
for x, y in zip(atlanta['year'], atlanta['avg_weekly_wage']):
    ax.annotate(f'${y:,.0f}', (x, y), xytext=(0, 9), textcoords='offset points', ha='center', fontsize=9)
plt.tight_layout(); plt.show()
print(f"Nominal change, 2021–2025: {(atlanta['avg_weekly_wage'].iloc[-1] / atlanta['avg_weekly_wage'].iloc[0] - 1) * 100:.1f}%.")
```

## Source and interpretation

Source: U.S. Bureau of Labor Statistics Quarterly Census of Employment and Wages (QCEW), Q4 county data. The project aggregates the counties included in its fixed 29-county Atlanta metro definition. These indicators describe the regional business and labor-market base; they do not measure company attraction directly or establish that incentives caused any change.
